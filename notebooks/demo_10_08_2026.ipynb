```python
import pandas as pd
```

```python
votes = pd.read_csv('https://voteview.com/static/data/out/votes/HS119_votes.csv')
```

```python
votes = votes[['chamber', 'rollnumber', 'icpsr', 'cast_code']]
```

```python
votes_compare = pd.merge(votes, votes, on=['chamber', 'rollnumber']).rename({'icpsr_x': 'icpsr', 'icpsr_y': 'comparison_member'}, axis=1)
```

```python
votes_compare
```

```python
votes_compare['agree'] = votes_compare['cast_code_x'] == votes_compare['cast_code_y']
votes_compare = votes_compare.query("icpsr != comparison_member")
```

```python
votes_compare
```

```python
votes_compare.groupby(['icpsr', 'comparison_member']).agg({'agree': 'mean'}).reset_index()
```
