There are five pieces of information you need to navigate any API and its poor documentation:

1. How you get an API key (or another credential) to get access to the data (check)

2. Root: the beginning of the URL for any endpoint contained within the API

A URL: `root / endpoint ? parameters`

3. Endpoint: the middle of the URL, specifies which data store on the API to access

4. Parameters: ways to control what exactly is returned when you access an endpoint. Two kinds of parameter:

* Query parameters (appear after the ?, but requests formats this for you using the `params` argument)
* Path parameters -- supposed to be entered as part of the endpoint itself

5. How to supply your API key back to the API to prove that you are you

```python
params['limit'] = '250'
```

```python
r = requests.get(root + endpoint, params=params, headers=headers)
myjson = r.json()
congress_df = pd.json_normalize(myjson, record_path=['members'])
congress_df
```

```python
i = 0
still_members = True
congress_json = []

while still_members:
    params['offset'] = i
    r = requests.get(root + endpoint, params=params, headers=headers)
    myjson = r.json()
    still_members = len(myjson['members']) > 0
    congress_json.extend(myjson['members'])

    i += 250

congress_df = pd.json_normalize(congress_json)
```

```python
macguire_df = congress_df.query('name.str.contains("McGuire") & district == 5')
```

```python
bioguideId = macguire_df.reset_index(drop=True)['bioguideId'][0]
```

```python
def get_sponsored_legislation(bioguideId):
    endpoint = f'/member/{bioguideId}/sponsored-legislation'
    params = {
        r'format': 'json',
        'currentMember': 'true',
        'api_key': congresskey
    }
    r = requests.get(root + endpoint, headers = headers, params = params)
    return r.json()['sponsoredLegislation']
```

```python
get_sponsored_legislation(bioguideId)
```

```python
def get_all_cosponsors(bioguideId):
    bills = get_sponsored_legislation(bioguideId)

    cosponsors = set()
    for bill in bills:
        cosponsors.update(get_bill_cosponsors(bill))

    
    return cosponsors
        

def get_bill_cosponsors(bill):
    params = {
        r'format': 'json',
        'currentMember': 'true',
        'api_key': congresskey
    }
    r = requests.get(bill['url'].split("?")[0] + "/cosponsors", headers = headers, params = params)

    cosponsors = [i['fullName'] for i in r.json()['cosponsors']]
    return cosponsors
```
